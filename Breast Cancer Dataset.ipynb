#### Boxplot of mean texture

```python
#boxplot to visualize the distribution of `mean texture`. Are there any outliers in thedata? What is the interquartile range (IQR)?
plt.figure(figsize=(10,5))
plt.boxplot(df_cancer['mean texture'])
plt.title('Boxplot of mean texture')
plt.show()
```

```python
Q1 = df_cancer['mean texture'].quantile(0.25)
Q3 = df_cancer['mean texture'].quantile(0.75)
IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_cancer[(df_cancer['mean texture'] < lower_bound) |
                     (df_cancer['mean texture'] > upper_bound)]

print("Number of outliers:", len(outliers))
```

```python
print(outliers['mean texture'])
```

A boxplot was created to visualize the distribution of the mean texture feature.

```python
#Create a heatmap to visualize the correlation matrix of the dataset. 
#Which features are most strongly correlated with each other? How might this affect your analysis?
plt.figure(figsize=(18, 15))
correlation_matrix=df_cancer.corr()
sns.heatmap(correlation_matrix,annot=True,cmap='viridis')
```

A correlation heatmap was created to examine the relationships among the features in the breast cancer dataset. The heatmap reveals several strong positive correlations between features that measure similar characteristics of tumors.

The **strongest positive correlations are observed among mean radius, mean perimeter, and mean area**.These features are highly related because they all describe the size of a tumor. As the radius of a tumor increases, its perimeter and area also increase.

Similarly, features such as radius error, perimeter error, and area error tend to be strongly correlated with one another.
The heatmap also shows that **mean compactness, mean concavity, and mean concave points are strongly positively correlated**. These features describe the shape and structural characteristics of the tumor and therefore tend to increase or decrease together.

These strong correlations indicate that some features contain similar information. For example, knowing the radius of a tumor already provides a good idea about its perimeter and area. Therefore, these features may be somewhat redundant.This can affect analysis because highly correlated features may not provide much additional information when used together. In machine learning models, this phenomenon is known as multicollinearity. While it does not necessarily reduce prediction accuracy, it can make model interpretation more difficult and increase the complexity of the model.

Overall, the heatmap reveals that many tumor size and shape features are closely related. The strongest relationships are observed among mean radius, mean perimeter, and mean area, suggesting that these features are important indicators of tumor characteristics and may play a significant role in distinguishing between benign and malignant tumors

### 5. Feature Engineering

```python
#Create a new feature called `area_to_perimeter_ratio` by dividing `mean area` by `mean perimeter`. 
#How does this new feature differ between benign and malignant cases?
df_cancer['area_to_perimeter_ratio']=df_cancer['mean area']/df_cancer['mean perimeter']
```

```python
df_cancer['area_to_perimeter_ratio']
```

```python
df_cancer.groupby('target')['area_to_perimeter_ratio'].mean()
```

A new feature, area_to_perimeter_ratio, was created by dividing the tumor's mean area by its mean perimeter.
The **average area_to_perimeter_ratio for malignant tumors is 8.21**, whereas the **average area_to_perimeter_ratio for benign tumors is 5.80.** 
Malignant tumors have a significantly higher average area-to-perimeter ratio than benign tumors. This indicates that **malignant tumors generally have a larger area relative to their perimeter than benign tumors**.

This result suggests that malignant tumors are generally larger in size. While both area and perimeter increase as a tumor grows, the area increases more rapidly than the perimeter. As a result, larger tumors tend to have higher area-to-perimeter ratios.

The difference between the average area_to_perimeter_ratio values of malignant and benign tumors indicates that this  feature captures meaningful information about tumor characteristics. An important advantage of this feature is that it combines information from two related measurements, area and perimeter, into a single measure, making it easier to distinguish between the two classes. While area and perimeter individually provide information about tumor size, the ratio captures the relationship between them.

Therefore, the area_to_perimeter_ratio appears to be a useful predictive feature.The feature  can be useful for cancer classification when combined with other predictive features. **Tumors with higher ratio values are more likely to be malignant, while tumors with lower ratio values are more likely to be benign.**

```python
# creating radius_to_texture_ratio
df_cancer['radius_to_texture_ratio'] = (
    df_cancer['mean radius'] / df_cancer['mean texture']
)
```

```python
df_cancer.groupby('target')['radius_to_texture_ratio'].mean()
```

A new feature called **radius_to_texture_ratio was created by dividing mean radius by mean texture**. This feature combines information about tumor size and texture into a single measurement.

The analysis showed that **malignant tumors have an average ratio of 0.831**, while **benign tumors have an average ratio of 0.710**. Although the difference is smaller than that observed for the area_to_perimeter_ratio, malignant tumors still tend to have higher values .

The higher ratio indicates that tumor size increases more than texture in malignant cases. Therefore, the relationship between radius and texture appears to differ between the two classes.The feature provides  an interaction between two characteristics of a tumor rather than examining them independently.

However there is only **small difference between average values of malignant and benign tumors indicates that radius_to_texture_ratio provides only moderate separability between the two classes.** So it is not strong predictor by itself but when combining with other features it can distinguish classes.
